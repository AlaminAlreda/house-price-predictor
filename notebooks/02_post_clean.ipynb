```python
import pandas as pd
df = pd.read_excel("../data/cleaned_houses.xlsx")
```

```python
# Inspect unique location names
print("Number of unique locations:", df["Location"].nunique())

for location in sorted(df["Location"].unique()):
    print(repr(location))
```

```python
def fix_encoding(value):
    if not isinstance(value, str):
        return value

    try:
        return value.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return value


for column in ["Address", "Location"]:
    df[column] = df[column].apply(fix_encoding)
```

```python
def fix_remaining_encoding(value):
    if not isinstance(value, str):
        return value

    replacements = {
        "Ã–": "Ö",
        "Ã¶": "ö",
        "Ã…": "Å",
        "Ã¥": "å",
        "Ã„": "Ä",
        "Ã¤": "ä",
        "Ã„": "Ä",
        #"Ã¼": "ü",
        "Ã©": "é",
        #"Â²": "²",
    }

    for wrong, correct in replacements.items():
        value = value.replace(wrong, correct)

    return value


for column in ["Address", "Location"]:
    df[column] = df[column].apply(fix_remaining_encoding)
```

```python
for column in ["Address", "Location"]:
    print(f"\nRemaining encoding issues in {column}:")

    for value in df[column].dropna().unique():
        if "Ã" in value or "Â" in value:
            print(repr(value))
```

```python
print("Number of unique locations:", df["Location"].nunique())

location_counts = df["Location"].value_counts()

for location in sorted(location_counts.index):
    print(f"{location!r}: {location_counts[location]} occurrences")
```
