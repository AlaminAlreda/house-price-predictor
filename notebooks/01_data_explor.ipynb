## Find how much data is missing

```python
# Drop the "Total Area" column
df = df.drop(columns=["Total Area", "Address"])

# Count rows that have at least one missing value
rows_with_missing = df.isna().any(axis=1).sum()

# Calculate the percentage of rows with at least one missing value
percentage_missing = (rows_with_missing / len(df)) * 100

print(f"Rows with at least one missing value: {rows_with_missing}")
print(f"Percentage of total rows: {percentage_missing:.2f}%")
```
